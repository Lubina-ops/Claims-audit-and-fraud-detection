```python
# =====================================================
# STANDARDIZE COLUMN NAMES
# =====================================================

manual.columns = manual.columns.str.strip()
ledger.columns = ledger.columns.str.strip()
claims.columns = claims.columns.str.strip()
```

```python
# =====================================================
# HELPER FUNCTIONS
# =====================================================

def clean_text(series):
    return (
        series.astype(str)
        .str.strip()
        .str.upper()
        .str.replace(r"\s+", " ", regex=True)
    )

def clean_id(series):
    return (
        series.astype(str)
        .str.strip()
        .str.replace(".0", "", regex=False)
    )
```

```python
# =====================================================
# CLEAN MANUAL PASS FILE
# =====================================================

manual["ParticipantName_clean"] = clean_text(manual["ParticipantName"])
manual["PromotionName_clean"] = clean_text(manual["PromotionName"])

manual["PromotionID_clean"] = clean_id(manual["PromotionID"])
manual["EntityID_clean"] = clean_id(manual["EntityID"])
manual["PSPS_clean"] = clean_text(manual["PSPS"])

manual["PointsPaid"] = pd.to_numeric(
    manual["PointsPaid"],
    errors="coerce"
)

manual["FulfillmentDate"] = pd.to_datetime(
    manual["FulfillmentDate"],
    errors="coerce"
)
```

```python
import pandas as pd
import numpy as np
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter


# ==========================================================
# CONFIGURATION
# ==========================================================

OUTPUT_FILE = "Business_Insights_Report_v3.xlsx"

CLAIM_ITEM_KEY = "EntityClaimItemEID"
CLAIM_KEY = "ClaimEId"
LEDGER_KEY = "LedgerId"

DATE_COLUMNS = [
    "ClaimSubmittedDate",
    "AwardDate",
    "FulfillmentDate"
]

CLAIM_SIDE_COLUMNS = [
    "EntityClaimItemEID",
    "ClaimEId",
    "ParticipantId",
    "ParticipantName",
    "EntityId",
    "PSPS",
    "PromotionName_claim",
    "ClaimItemStatus",
    "ClaimSubmittedDate"
]

LEDGER_SIDE_COLUMNS = [
    "EntityClaimItemEID",
    "LedgerId",
    "AwardDate",
    "AwardAmount"
]


# ==========================================================
# 1. VALIDATE REQUIRED COLUMNS
# ==========================================================

required_columns = list(
    set(
        CLAIM_SIDE_COLUMNS
        + LEDGER_SIDE_COLUMNS
    )
)

missing_columns = [
    col for col in required_columns
    if col not in full_chain.columns
]

if missing_columns:
    raise ValueError(
        "The following required columns are missing from full_chain: "
        + ", ".join(sorted(missing_columns))
    )


# ==========================================================
# 2. CREATE WORKING COPY
# ==========================================================

df = full_chain.copy()

input_row_count = len(df)


# ==========================================================
# 3. DATE CLEANUP
# ==========================================================

for col in DATE_COLUMNS:
    if col in df.columns:
        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )


# ==========================================================
# 4. STANDARDIZE BLANK IDENTIFIERS
# ==========================================================

identifier_columns = [
    "EntityClaimItemEID",
    "ClaimEId",
    "ParticipantId",
    "EntityId",
    "PSPS",
    "LedgerId"
]

for col in identifier_columns:
    if col in df.columns:
        df[col] = df[col].replace(
            r"^\s*$",
            np.nan,
            regex=True
        )


# ==========================================================
# 5. DATA QUALITY COUNTS BEFORE BUILDING REPORT
# ==========================================================

missing_claim_item_id_rows = (
    df[CLAIM_ITEM_KEY].isna().sum()
)

rows_with_claim_item_id = (
    df[CLAIM_ITEM_KEY].notna().sum()
)

rows_with_ledger_id = (
    df[LEDGER_KEY].notna().sum()
)

unique_claim_items_in_source = (
    df[CLAIM_ITEM_KEY]
    .dropna()
    .nunique()
)

unique_claims_in_source = (
    df[CLAIM_KEY]
    .dropna()
    .nunique()
)

unique_ledger_ids_in_source = (
    df[LEDGER_KEY]
    .dropna()
    .nunique()
)


# ==========================================================
# 6. CLAIM-SIDE CONSISTENCY CHECK
#
# Confirm that claim-side attributes are consistent within
# each EntityClaimItemEID before reducing to one row.
# ==========================================================

claim_consistency = (
    df[
        [
            "EntityClaimItemEID",
            "ClaimEId",
            "ParticipantId",
            "ParticipantName",
            "EntityId",
            "PSPS",
            "PromotionName_claim",
            "ClaimItemStatus",
            "ClaimSubmittedDate"
        ]
    ]
    .dropna(subset=["EntityClaimItemEID"])
    .groupby(
        "EntityClaimItemEID",
        dropna=False
    )
    .agg(
        ClaimEIdVersions=(
            "ClaimEId",
            lambda x: x.dropna().nunique()
        ),
        ParticipantIdVersions=(
            "ParticipantId",
            lambda x: x.dropna().nunique()
        ),
        ParticipantNameVersions=(
            "ParticipantName",
            lambda x: x.dropna().nunique()
        ),
        EntityIdVersions=(
            "EntityId",
            lambda x: x.dropna().nunique()
        ),
        PSPSVersions=(
            "PSPS",
            lambda x: x.dropna().nunique()
        ),
        PromotionVersions=(
            "PromotionName_claim",
            lambda x: x.dropna().nunique()
        ),
        StatusVersions=(
            "ClaimItemStatus",
            lambda x: x.dropna().nunique()
        ),
        SubmittedDateVersions=(
            "ClaimSubmittedDate",
            lambda x: x.dropna().nunique()
        )
    )
    .reset_index()
)

version_columns = [
    col for col in claim_consistency.columns
    if col.endswith("Versions")
]

claim_attribute_conflicts = (
    claim_consistency[
        claim_consistency[version_columns]
        .gt(1)
        .any(axis=1)
    ]
    .copy()
)


# ==========================================================
# 7. CREATE CLAIM-ITEM BASE TABLE
#
# One row per EntityClaimItemEID.
#
# Claim-side fields are taken only from claim data.
# sort_values creates deterministic behavior before
# drop_duplicates.
# ==========================================================

claim_item_base = (
    df[CLAIM_SIDE_COLUMNS]
    .dropna(subset=["EntityClaimItemEID"])
    .sort_values(
        by=[
            "EntityClaimItemEID",
            "ClaimSubmittedDate"
        ],
        na_position="last"
    )
    .drop_duplicates(
        subset=["EntityClaimItemEID"],
        keep="first"
    )
    .rename(
        columns={
            "PromotionName_claim": "PromotionName"
        }
    )
    .reset_index(drop=True)
)


# ==========================================================
# 8. CREATE DISTINCT CLAIM-ITEM / LEDGER CANDIDATE BRIDGE
#
# One row per:
# EntityClaimItemEID + LedgerId + AwardDate + AwardAmount
#
# This makes candidate relationships explicit.
# ==========================================================

candidate_bridge = (
    df[LEDGER_SIDE_COLUMNS]
    .dropna(
        subset=[
            "EntityClaimItemEID",
            "LedgerId"
        ]
    )
    .drop_duplicates()
    .reset_index(drop=True)
)


# ==========================================================
# 9. CANDIDATE-LEVEL SUMMARY
#
# Count ledger IDs, award dates, and award amounts separately.
# This prevents an arbitrary AwardDate or AwardAmount from
# being selected for unresolved claim items.
# ==========================================================

candidate_summary = (
    candidate_bridge
    .groupby(
        "EntityClaimItemEID",
        dropna=False
    )
    .agg(
        DistinctLedgerIds=(
            "LedgerId",
            lambda x: x.dropna().nunique()
        ),
        DistinctAwardDates=(
            "AwardDate",
            lambda x: x.dropna().nunique()
        ),
        DistinctAwardAmounts=(
            "AwardAmount",
            lambda x: x.dropna().nunique()
        ),
        CandidateRows=(
            "LedgerId",
            "size"
        )
    )
    .reset_index()
)


# ==========================================================
# 10. IDENTIFY SINGLE-CANDIDATE ITEMS
# ==========================================================

single_candidate_ids = candidate_summary.loc[
    candidate_summary["DistinctLedgerIds"] == 1,
    "EntityClaimItemEID"
]


# ==========================================================
# 11. CREATE CONFIRMED SINGLE-CANDIDATE AWARD TABLE
#
# Award data is retained only where:
# 1. There is exactly one LedgerId
# 2. There is no conflict in AwardDate
# 3. There is no conflict in AwardAmount
# ==========================================================

single_candidate_bridge = (
    candidate_bridge[
        candidate_bridge["EntityClaimItemEID"]
        .isin(single_candidate_ids)
    ]
    .copy()
)

single_candidate_awards = (
    single_candidate_bridge
    .sort_values(
        by=[
            "EntityClaimItemEID",
            "AwardDate",
            "LedgerId"
        ],
        na_position="last"
    )
    .groupby(
        "EntityClaimItemEID",
        dropna=False
    )
    .agg(
        ConfirmedLedgerId=(
            "LedgerId",
            "first"
        ),
        ConfirmedAwardDate=(
            "AwardDate",
            "first"
        ),
        ConfirmedAwardAmount=(
            "AwardAmount",
            "first"
        ),
        SingleCandidateAwardDateVersions=(
            "AwardDate",
            lambda x: x.dropna().nunique()
        ),
        SingleCandidateAwardAmountVersions=(
            "AwardAmount",
            lambda x: x.dropna().nunique()
        )
    )
    .reset_index()
)


# ==========================================================
# 12. REMOVE CONFLICTING AWARD DETAILS
#
# Even when one LedgerId exists, conflicting dates or amounts
# should not be treated as confirmed.
# ==========================================================

single_candidate_awards[
    "SingleCandidateAwardConflict"
] = np.where(
    (
        single_candidate_awards[
            "SingleCandidateAwardDateVersions"
        ] > 1
    )
    |
    (
        single_candidate_awards[
            "SingleCandidateAwardAmountVersions"
        ] > 1
    ),
    "Yes",
    "No"
)

award_conflict_mask = (
    single_candidate_awards[
        "SingleCandidateAwardConflict"
    ] == "Yes"
)

single_candidate_awards.loc[
    award_conflict_mask,
    [
        "ConfirmedAwardDate",
        "ConfirmedAwardAmount"
    ]
] = np.nan


# ==========================================================
# 13. BUILD FINAL CLAIM-ITEM ANALYTICAL TABLE
# ==========================================================

claim_item_summary = (
    claim_item_base
    .merge(
        candidate_summary,
        on="EntityClaimItemEID",
        how="left",
        validate="one_to_one"
    )
    .merge(
        single_candidate_awards,
        on="EntityClaimItemEID",
        how="left",
        validate="one_to_one"
    )
)


# ==========================================================
# 14. FILL NO-CANDIDATE VALUES
# ==========================================================

count_columns = [
    "DistinctLedgerIds",
    "DistinctAwardDates",
    "DistinctAwardAmounts",
    "CandidateRows"
]

for col in count_columns:
    claim_item_summary[col] = (
        claim_item_summary[col]
        .fillna(0)
        .astype(int)
    )

claim_item_summary[
    "SingleCandidateAwardConflict"
] = (
    claim_item_summary[
        "SingleCandidateAwardConflict"
    ]
    .fillna("Not Applicable")
)


# ==========================================================
# 15. CANDIDATE MATCH STATUS
#
# These names describe candidate relationships only.
# They do not imply confirmed payouts.
# ==========================================================

claim_item_summary[
    "CandidateMatchStatus"
] = np.select(
    [
        claim_item_summary[
            "DistinctLedgerIds"
        ] == 0,

        claim_item_summary[
            "DistinctLedgerIds"
        ] == 1,

        claim_item_summary[
            "DistinctLedgerIds"
        ] > 1
    ],
    [
        "No Candidate Match",
        "Single Candidate Match",
        "Multiple Candidate Matches"
    ],
    default="Unknown"
)


# ==========================================================
# 16. AWARD RESOLUTION STATUS
# ==========================================================

claim_item_summary[
    "AwardResolutionStatus"
] = np.select(
    [
        claim_item_summary[
            "DistinctLedgerIds"
        ] == 0,

        claim_item_summary[
            "DistinctLedgerIds"
        ] > 1,

        (
            claim_item_summary[
                "DistinctLedgerIds"
            ] == 1
        )
        &
        (
            claim_item_summary[
                "SingleCandidateAwardConflict"
            ] == "Yes"
        ),

        (
            claim_item_summary[
                "DistinctLedgerIds"
            ] == 1
        )
        &
        (
            claim_item_summary[
                "ConfirmedAwardDate"
            ].isna()
        ),

        (
            claim_item_summary[
                "DistinctLedgerIds"
            ] == 1
        )
        &
        (
            claim_item_summary[
                "ConfirmedAwardDate"
            ].notna()
        )
    ],
    [
        "Unmatched",
        "Unresolved Multiple Candidates",
        "Single Candidate With Conflicting Award Data",
        "Single Candidate Missing Award Date",
        "Resolved Single Candidate"
    ],
    default="Review Required"
)


# ==========================================================
# 17. CONFIRMED CLAIM-TO-AWARD CYCLE TIME
#
# Cycle time is calculated only from the confirmed award date.
# Multiple-candidate records are not assigned arbitrary dates.
# ==========================================================

claim_item_summary[
    "Days_Claim_to_Award"
] = (
    claim_item_summary[
        "ConfirmedAwardDate"
    ]
    -
    claim_item_summary[
        "ClaimSubmittedDate"
    ]
).dt.days


# ==========================================================
# 18. CYCLE-TIME DATA QUALITY STATUS
# ==========================================================

claim_item_summary[
    "CycleTimeStatus"
] = np.select(
    [
        claim_item_summary[
            "AwardResolutionStatus"
        ] != "Resolved Single Candidate",

        claim_item_summary[
            "ClaimSubmittedDate"
        ].isna(),

        claim_item_summary[
            "ConfirmedAwardDate"
        ].isna(),

        claim_item_summary[
            "Days_Claim_to_Award"
        ] < 0,

        claim_item_summary[
            "Days_Claim_to_Award"
        ] == 0,

        claim_item_summary[
            "Days_Claim_to_Award"
        ] > 0
    ],
    [
        "Not Eligible - Award Unresolved",
        "Not Eligible - Missing Claim Date",
        "Not Eligible - Missing Award Date",
        "Invalid Negative Duration",
        "Valid Same-Day Award",
        "Valid Positive Duration"
    ],
    default="Review Required"
)


# ==========================================================
# 19. VALID CYCLE-TIME DATASET
#
# Retain only zero or positive confirmed cycle times.
# Negative durations remain visible in the exception report.
# ==========================================================

valid_cycle_time = (
    claim_item_summary[
        claim_item_summary[
            "CycleTimeStatus"
        ].isin(
            [
                "Valid Same-Day Award",
                "Valid Positive Duration"
            ]
        )
    ]
    .copy()
)


# ==========================================================
# 20. CYCLE-TIME SUMMARY
# ==========================================================

if not valid_cycle_time.empty:

    cycle_time_series = (
        valid_cycle_time[
            "Days_Claim_to_Award"
        ]
        .dropna()
    )

    cycle_summary = pd.DataFrame(
        {
            "Metric": [
                "Eligible Claim Items",
                "Mean Days",
                "Median Days",
                "25th Percentile",
                "75th Percentile",
                "90th Percentile",
                "95th Percentile",
                "Minimum Days",
                "Maximum Days"
            ],
            "Value": [
                cycle_time_series.count(),
                cycle_time_series.mean(),
                cycle_time_series.median(),
                cycle_time_series.quantile(0.25),
                cycle_time_series.quantile(0.75),
                cycle_time_series.quantile(0.90),
                cycle_time_series.quantile(0.95),
                cycle_time_series.min(),
                cycle_time_series.max()
            ]
        }
    )

else:

    cycle_summary = pd.DataFrame(
        {
            "Metric": [
                "Eligible Claim Items"
            ],
            "Value": [
                0
            ]
        }
    )


# ==========================================================
# 21. AWARD-TIME BUCKETS
#
# These are claim-to-award buckets, not fulfillment buckets.
# ==========================================================

claim_item_summary[
    "AwardTimeBucket"
] = pd.cut(
    claim_item_summary[
        "Days_Claim_to_Award"
    ],
    bins=[
        -0.001,
        7,
        30,
        60,
        90,
        np.inf
    ],
    labels=[
        "0-7 Days",
        "8-30 Days",
        "31-60 Days",
        "61-90 Days",
        "90+ Days"
    ],
    include_lowest=True
)

award_time_distribution = (
    claim_item_summary[
        "AwardTimeBucket"
    ]
    .value_counts(
        sort=False,
        dropna=False
    )
    .rename_axis("AwardTimeBucket")
    .reset_index(name="ClaimItems")
)

award_time_distribution[
    "AwardTimeBucket"
] = (
    award_time_distribution[
        "AwardTimeBucket"
    ]
    .astype(str)
    .replace(
        "nan",
        "Not Eligible or Unresolved"
    )
)

award_time_distribution[
    "PercentOfAllClaimItems"
] = (
    award_time_distribution[
        "ClaimItems"
    ]
    /
    len(claim_item_summary)
    * 100
).round(2)


# ==========================================================
# 22. PROMOTION DELAY ANALYSIS
#
# TotalClaimItems shows total population.
# CycleTimeEligibleItems shows the actual denominator used.
# ==========================================================

promo_total = (
    claim_item_summary
    .groupby(
        "PromotionName",
        dropna=False
    )
    .agg(
        TotalClaimItems=(
            "EntityClaimItemEID",
            "nunique"
        )
    )
    .reset_index()
)

promo_cycle = (
    valid_cycle_time
    .groupby(
        "PromotionName",
        dropna=False
    )
    .agg(
        CycleTimeEligibleItems=(
            "EntityClaimItemEID",
            "nunique"
        ),
        AvgDays=(
            "Days_Claim_to_Award",
            "mean"
        ),
        MedianDays=(
            "Days_Claim_to_Award",
            "median"
        ),
        P90Days=(
            "Days_Claim_to_Award",
            lambda x: x.quantile(0.90)
        ),
        MaxDays=(
            "Days_Claim_to_Award",
            "max"
        )
    )
    .reset_index()
)

promo_delay = (
    promo_total
    .merge(
        promo_cycle,
        on="PromotionName",
        how="left",
        validate="one_to_one"
    )
)

promo_delay[
    "CycleTimeEligibleItems"
] = (
    promo_delay[
        "CycleTimeEligibleItems"
    ]
    .fillna(0)
    .astype(int)
)

promo_delay[
    "CycleTimeCoveragePct"
] = (
    promo_delay[
        "CycleTimeEligibleItems"
    ]
    /
    promo_delay[
        "TotalClaimItems"
    ]
    * 100
).round(2)

promo_delay[
    "SampleSizeFlag"
] = np.where(
    promo_delay[
        "CycleTimeEligibleItems"
    ] >= 20,
    "Sufficient for Review",
    "Small Sample - Use Caution"
)

promo_delay = promo_delay.sort_values(
    by=[
        "MedianDays",
        "CycleTimeEligibleItems"
    ],
    ascending=[
        False,
        False
    ],
    na_position="last"
)


# ==========================================================
# 23. ENTITY DELAY ANALYSIS
# ==========================================================

entity_total = (
    claim_item_summary
    .groupby(
        "EntityId",
        dropna=False
    )
    .agg(
        TotalClaimItems=(
            "EntityClaimItemEID",
            "nunique"
        )
    )
    .reset_index()
)

entity_cycle = (
    valid_cycle_time
    .groupby(
        "EntityId",
        dropna=False
    )
    .agg(
        CycleTimeEligibleItems=(
            "EntityClaimItemEID",
            "nunique"
        ),
        AvgDays=(
            "Days_Claim_to_Award",
            "mean"
        ),
        MedianDays=(
            "Days_Claim_to_Award",
            "median"
        ),
        P90Days=(
            "Days_Claim_to_Award",
            lambda x: x.quantile(0.90)
        ),
        MaxDays=(
            "Days_Claim_to_Award",
            "max"
        )
    )
    .reset_index()
)

entity_delay = (
    entity_total
    .merge(
        entity_cycle,
        on="EntityId",
        how="left",
        validate="one_to_one"
    )
)

entity_delay[
    "CycleTimeEligibleItems"
] = (
    entity_delay[
        "CycleTimeEligibleItems"
    ]
    .fillna(0)
    .astype(int)
)

entity_delay[
    "CycleTimeCoveragePct"
] = (
    entity_delay[
        "CycleTimeEligibleItems"
    ]
    /
    entity_delay[
        "TotalClaimItems"
    ]
    * 100
).round(2)

entity_delay[
    "SampleSizeFlag"
] = np.where(
    entity_delay[
        "CycleTimeEligibleItems"
    ] >= 20,
    "Sufficient for Review",
    "Small Sample - Use Caution"
)

entity_delay = entity_delay.sort_values(
    by=[
        "MedianDays",
        "CycleTimeEligibleItems"
    ],
    ascending=[
        False,
        False
    ],
    na_position="last"
)


# ==========================================================
# 24. CANDIDATE MATCH DISTRIBUTION
# ==========================================================

candidate_distribution = (
    claim_item_summary[
        "CandidateMatchStatus"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "CandidateMatchStatus"
    )
    .reset_index(
        name="ClaimItems"
    )
)

candidate_distribution[
    "Percent"
] = (
    candidate_distribution[
        "ClaimItems"
    ]
    /
    len(claim_item_summary)
    * 100
).round(2)


# ==========================================================
# 25. AWARD RESOLUTION DISTRIBUTION
# ==========================================================

award_resolution_distribution = (
    claim_item_summary[
        "AwardResolutionStatus"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "AwardResolutionStatus"
    )
    .reset_index(
        name="ClaimItems"
    )
)

award_resolution_distribution[
    "Percent"
] = (
    award_resolution_distribution[
        "ClaimItems"
    ]
    /
    len(claim_item_summary)
    * 100
).round(2)


# ==========================================================
# 26. MULTIPLE CANDIDATE MATCHES
#
# This does not claim that multiple payouts occurred.
# ==========================================================

multiple_candidate_matches = (
    claim_item_summary[
        claim_item_summary[
            "DistinctLedgerIds"
        ] > 1
    ]
    .sort_values(
        by=[
            "DistinctLedgerIds",
            "EntityClaimItemEID"
        ],
        ascending=[
            False,
            True
        ]
    )
    .copy()
)


# ==========================================================
# 27. MULTIPLE-CANDIDATE RATE BY PROMOTION
# ==========================================================

promo_candidate_rates = (
    claim_item_summary
    .groupby(
        "PromotionName",
        dropna=False
    )
    .agg(
        TotalClaimItems=(
            "EntityClaimItemEID",
            "nunique"
        ),
        NoCandidateItems=(
            "CandidateMatchStatus",
            lambda x: (
                x == "No Candidate Match"
            ).sum()
        ),
        SingleCandidateItems=(
            "CandidateMatchStatus",
            lambda x: (
                x == "Single Candidate Match"
            ).sum()
        ),
        MultipleCandidateItems=(
            "CandidateMatchStatus",
            lambda x: (
                x == "Multiple Candidate Matches"
            ).sum()
        )
    )
    .reset_index()
)

promo_candidate_rates[
    "MultipleCandidateRatePct"
] = (
    promo_candidate_rates[
        "MultipleCandidateItems"
    ]
    /
    promo_candidate_rates[
        "TotalClaimItems"
    ]
    * 100
).round(2)

promo_candidate_rates[
    "NoCandidateRatePct"
] = (
    promo_candidate_rates[
        "NoCandidateItems"
    ]
    /
    promo_candidate_rates[
        "TotalClaimItems"
    ]
    * 100
).round(2)

promo_candidate_rates[
    "SampleSizeFlag"
] = np.where(
    promo_candidate_rates[
        "TotalClaimItems"
    ] >= 20,
    "Sufficient for Review",
    "Small Sample - Use Caution"
)

promo_candidate_rates = (
    promo_candidate_rates
    .sort_values(
        by=[
            "MultipleCandidateRatePct",
            "TotalClaimItems"
        ],
        ascending=[
            False,
            False
        ]
    )
)


# ==========================================================
# 28. CONFIRMED AWARD ANALYSIS
#
# Only resolved single-candidate items contain confirmed
# award details.
# ==========================================================

confirmed_award_analysis = (
    claim_item_summary[
        [
            "EntityClaimItemEID",
            "ClaimEId",
            "ParticipantId",
            "EntityId",
            "PSPS",
            "PromotionName",
            "ClaimItemStatus",
            "ClaimSubmittedDate",
            "ConfirmedLedgerId",
            "ConfirmedAwardDate",
            "ConfirmedAwardAmount",
            "Days_Claim_to_Award",
            "CycleTimeStatus",
            "AwardResolutionStatus"
        ]
    ]
    .copy()
)


# ==========================================================
# 29. PARTICIPANT ANALYSIS
# ==========================================================

participant_analysis = (
    claim_item_summary
    .groupby(
        "ParticipantId",
        dropna=False
    )
    .agg(
        ClaimItems=(
            "EntityClaimItemEID",
            "nunique"
        ),
        Claims=(
            "ClaimEId",
            "nunique"
        ),
        NoCandidateItems=(
            "CandidateMatchStatus",
            lambda x: (
                x == "No Candidate Match"
            ).sum()
        ),
        SingleCandidateItems=(
            "CandidateMatchStatus",
            lambda x: (
                x == "Single Candidate Match"
            ).sum()
        ),
        MultipleCandidateItems=(
            "CandidateMatchStatus",
            lambda x: (
                x == "Multiple Candidate Matches"
            ).sum()
        ),
        ResolvedAwardItems=(
            "AwardResolutionStatus",
            lambda x: (
                x == "Resolved Single Candidate"
            ).sum()
        )
    )
    .reset_index()
    .sort_values(
        "ClaimItems",
        ascending=False
    )
)


# ==========================================================
# 30. CLAIM ITEMS WITH NO CANDIDATE MATCH
# ==========================================================

no_candidate_items = (
    claim_item_summary[
        claim_item_summary[
            "CandidateMatchStatus"
        ] == "No Candidate Match"
    ]
    .copy()
)

no_candidate_summary = (
    no_candidate_items
    .groupby(
        "PromotionName",
        dropna=False
    )
    .agg(
        ClaimItems=(
            "EntityClaimItemEID",
            "nunique"
        ),
        Claims=(
            "ClaimEId",
            "nunique"
        ),
        Participants=(
            "ParticipantId",
            "nunique"
        )
    )
    .reset_index()
    .sort_values(
        "ClaimItems",
        ascending=False
    )
)


# ==========================================================
# 31. CYCLE-TIME EXCEPTIONS
# ==========================================================

cycle_time_exceptions = (
    claim_item_summary[
        ~claim_item_summary[
            "CycleTimeStatus"
        ].isin(
            [
                "Valid Same-Day Award",
                "Valid Positive Duration"
            ]
        )
    ]
    .copy()
    .sort_values(
        by=[
            "CycleTimeStatus",
            "EntityClaimItemEID"
        ]
    )
)


# ==========================================================
# 32. SINGLE-CANDIDATE AWARD CONFLICTS
# ==========================================================

single_candidate_conflicts = (
    claim_item_summary[
        claim_item_summary[
            "SingleCandidateAwardConflict"
        ] == "Yes"
    ]
    .copy()
)


# ==========================================================
# 33. RECONCILIATION CONTROLS
# ==========================================================

total_analytical_claim_items = len(
    claim_item_summary
)

no_candidate_count = (
    claim_item_summary[
        "CandidateMatchStatus"
    ]
    .eq("No Candidate Match")
    .sum()
)

single_candidate_count = (
    claim_item_summary[
        "CandidateMatchStatus"
    ]
    .eq("Single Candidate Match")
    .sum()
)

multiple_candidate_count = (
    claim_item_summary[
        "CandidateMatchStatus"
    ]
    .eq("Multiple Candidate Matches")
    .sum()
)

candidate_category_total = (
    no_candidate_count
    + single_candidate_count
    + multiple_candidate_count
)

resolved_award_count = (
    claim_item_summary[
        "AwardResolutionStatus"
    ]
    .eq("Resolved Single Candidate")
    .sum()
)

valid_cycle_time_count = len(
    valid_cycle_time
)

negative_cycle_time_count = (
    claim_item_summary[
        "Days_Claim_to_Award"
    ]
    .lt(0)
    .sum()
)

same_day_cycle_time_count = (
    claim_item_summary[
        "Days_Claim_to_Award"
    ]
    .eq(0)
    .sum()
)

reconciliation_controls = pd.DataFrame(
    {
        "Control": [
            "Input full_chain rows",
            "Rows with EntityClaimItemEID",
            "Rows missing EntityClaimItemEID",
            "Rows with LedgerId",
            "Unique claim items in source",
            "Claim items in analytical layer",
            "Claim-item count preserved",
            "Unique claims in source",
            "Distinct ledger IDs in source",
            "No candidate match items",
            "Single candidate match items",
            "Multiple candidate match items",
            "Candidate category total",
            "Candidate categories reconcile",
            "Resolved single-candidate awards",
            "Valid cycle-time items",
            "Same-day award items",
            "Negative cycle-time exceptions",
            "Claim attribute conflict items",
            "Single-candidate award conflict items"
        ],
        "Value": [
            input_row_count,
            rows_with_claim_item_id,
            missing_claim_item_id_rows,
            rows_with_ledger_id,
            unique_claim_items_in_source,
            total_analytical_claim_items,
            (
                unique_claim_items_in_source
                == total_analytical_claim_items
            ),
            unique_claims_in_source,
            unique_ledger_ids_in_source,
            no_candidate_count,
            single_candidate_count,
            multiple_candidate_count,
            candidate_category_total,
            (
                candidate_category_total
                == total_analytical_claim_items
            ),
            resolved_award_count,
            valid_cycle_time_count,
            same_day_cycle_time_count,
            negative_cycle_time_count,
            len(claim_attribute_conflicts),
            len(single_candidate_conflicts)
        ]
    }
)


# ==========================================================
# 34. EXECUTIVE KPIs
# ==========================================================

unique_participants = (
    claim_item_summary[
        "ParticipantId"
    ]
    .dropna()
    .nunique()
)

candidate_match_rate = (
    (
        single_candidate_count
        + multiple_candidate_count
    )
    /
    total_analytical_claim_items
    * 100
    if total_analytical_claim_items > 0
    else 0
)

no_candidate_rate = (
    no_candidate_count
    /
    total_analytical_claim_items
    * 100
    if total_analytical_claim_items > 0
    else 0
)

single_candidate_rate = (
    single_candidate_count
    /
    total_analytical_claim_items
    * 100
    if total_analytical_claim_items > 0
    else 0
)

multiple_candidate_rate = (
    multiple_candidate_count
    /
    total_analytical_claim_items
    * 100
    if total_analytical_claim_items > 0
    else 0
)

cycle_time_coverage_rate = (
    valid_cycle_time_count
    /
    total_analytical_claim_items
    * 100
    if total_analytical_claim_items > 0
    else 0
)

kpis = pd.DataFrame(
    {
        "Metric": [
            "Unique Claims",
            "Unique Claim Items",
            "Unique Participants",
            "Distinct Ledger IDs",
            "No Candidate Match Items",
            "Single Candidate Match Items",
            "Multiple Candidate Match Items",
            "Candidate Match Rate Pct",
            "No Candidate Rate Pct",
            "Single Candidate Rate Pct",
            "Multiple Candidate Rate Pct",
            "Resolved Single-Candidate Awards",
            "Valid Cycle-Time Items",
            "Cycle-Time Coverage Pct",
            "Negative Cycle-Time Exceptions",
            "Claim Attribute Conflict Items",
            "Single-Candidate Award Conflict Items"
        ],
        "Value": [
            claim_item_summary[
                "ClaimEId"
            ].nunique(),
            claim_item_summary[
                "EntityClaimItemEID"
            ].nunique(),
            unique_participants,
            unique_ledger_ids_in_source,
            no_candidate_count,
            single_candidate_count,
            multiple_candidate_count,
            round(candidate_match_rate, 2),
            round(no_candidate_rate, 2),
            round(single_candidate_rate, 2),
            round(multiple_candidate_rate, 2),
            resolved_award_count,
            valid_cycle_time_count,
            round(cycle_time_coverage_rate, 2),
            negative_cycle_time_count,
            len(claim_attribute_conflicts),
            len(single_candidate_conflicts)
        ]
    }
)


# ==========================================================
# 35. METHODOLOGY NOTES
# ==========================================================

methodology_notes = pd.DataFrame(
    {
        "Topic": [
            "Analytical Grain",
            "Candidate Definition",
            "No Candidate Match",
            "Single Candidate Match",
            "Multiple Candidate Matches",
            "Confirmed Award",
            "Cycle Time",
            "Negative Cycle Time",
            "Promotion and Entity Delays",
            "Sample Size",
            "Important Limitation"
        ],
        "Definition": [
            "One analytical row represents one EntityClaimItemEID.",
            "A candidate is a distinct LedgerId connected to an EntityClaimItemEID in full_chain.",
            "The claim item has zero distinct LedgerId values in full_chain.",
            "The claim item has exactly one distinct LedgerId in full_chain.",
            "The claim item has more than one distinct LedgerId in full_chain. This does not by itself prove multiple payouts.",
            "Award details are treated as confirmed only for a single-candidate item without conflicting award dates or award amounts.",
            "Days from ClaimSubmittedDate to ConfirmedAwardDate. Only resolved single-candidate records with non-negative durations are included in valid cycle-time summaries.",
            "A negative duration is retained as a data-quality exception and excluded from valid cycle-time summaries.",
            "Delay statistics use only valid cycle-time records. TotalClaimItems and CycleTimeEligibleItems are reported separately.",
            "Groups with fewer than 20 eligible cycle-time records are marked Small Sample - Use Caution.",
            "The accuracy of candidate relationships depends on the upstream joins used to create full_chain. This report validates and summarizes full_chain but does not independently prove that every upstream match is correct."
        ]
    }
)


# ==========================================================
# 36. EXPORT REPORT
# ==========================================================

with pd.ExcelWriter(
    OUTPUT_FILE,
    engine="openpyxl",
    datetime_format="mm/dd/yyyy",
    date_format="mm/dd/yyyy"
) as writer:

    kpis.to_excel(
        writer,
        sheet_name="Executive_KPIs",
        index=False
    )

    reconciliation_controls.to_excel(
        writer,
        sheet_name="Reconciliation_Controls",
        index=False
    )

    methodology_notes.to_excel(
        writer,
        sheet_name="Methodology_Notes",
        index=False
    )

    candidate_distribution.to_excel(
        writer,
        sheet_name="Candidate_Distribution",
        index=False
    )

    award_resolution_distribution.to_excel(
        writer,
        sheet_name="Award_Resolution",
        index=False
    )

    cycle_summary.to_excel(
        writer,
        sheet_name="Cycle_Time_Summary",
        index=False
    )

    promo_delay.to_excel(
        writer,
        sheet_name="Promotion_Delays",
        index=False
    )

    entity_delay.to_excel(
        writer,
        sheet_name="Entity_Delays",
        index=False
    )

    promo_candidate_rates.to_excel(
        writer,
        sheet_name="Promo_Candidate_Rates",
        index=False
    )

    multiple_candidate_matches.to_excel(
        writer,
        sheet_name="Multiple_Candidates",
        index=False
    )

    confirmed_award_analysis.to_excel(
        writer,
        sheet_name="Confirmed_Awards",
        index=False
    )

    award_time_distribution.to_excel(
        writer,
        sheet_name="Award_Time_Buckets",
        index=False
    )

    participant_analysis.to_excel(
        writer,
        sheet_name="Top_Participants",
        index=False
    )

    no_candidate_summary.to_excel(
        writer,
        sheet_name="No_Candidate_Summary",
        index=False
    )

    no_candidate_items.to_excel(
        writer,
        sheet_name="No_Candidate_Detail",
        index=False
    )

    cycle_time_exceptions.to_excel(
        writer,
        sheet_name="Cycle_Time_Exceptions",
        index=False
    )

    claim_attribute_conflicts.to_excel(
        writer,
        sheet_name="Claim_Attr_Conflicts",
        index=False
    )

    single_candidate_conflicts.to_excel(
        writer,
        sheet_name="Award_Data_Conflicts",
        index=False
    )

    candidate_bridge.to_excel(
        writer,
        sheet_name="Candidate_Bridge",
        index=False
    )

    claim_item_summary.to_excel(
        writer,
        sheet_name="Claim_Item_Detail",
        index=False
    )

    # ======================================================
    # 37. BASIC EXCEL FORMATTING
    # ======================================================

    workbook = writer.book

    header_fill = PatternFill(
        fill_type="solid",
        fgColor="1F4E78"
    )

    header_font = Font(
        color="FFFFFF",
        bold=True
    )

    exception_fill = PatternFill(
        fill_type="solid",
        fgColor="FFF2CC"
    )

    for worksheet in workbook.worksheets:

        worksheet.freeze_panes = "A2"
        worksheet.auto_filter.ref = worksheet.dimensions

        for cell in worksheet[1]:
            cell.fill = header_fill
            cell.font = header_font
            cell.alignment = Alignment(
                horizontal="left",
                vertical="center"
            )

        for row in worksheet.iter_rows():
            for cell in row:
                cell.alignment = Alignment(
                    horizontal="left",
                    vertical="top"
                )

        for column_cells in worksheet.columns:

            column_letter = get_column_letter(
                column_cells[0].column
            )

            max_length = 0

            for cell in column_cells:

                if cell.value is not None:
                    cell_length = len(
                        str(cell.value)
                    )

                    max_length = max(
                        max_length,
                        cell_length
                    )

            worksheet.column_dimensions[
                column_letter
            ].width = min(
                max(max_length + 2, 12),
                45
            )

        # Apply date display to date columns
        for cell in worksheet[1]:
            if cell.value in [
                "ClaimSubmittedDate",
                "AwardDate",
                "FulfillmentDate",
                "ConfirmedAwardDate"
            ]:

                date_column = cell.column

                for row_number in range(
                    2,
                    worksheet.max_row + 1
                ):
                    worksheet.cell(
                        row=row_number,
                        column=date_column
                    ).number_format = "mm/dd/yyyy"

        # Highlight exception sheets
        if worksheet.title in [
            "Cycle_Time_Exceptions",
            "Claim_Attr_Conflicts",
            "Award_Data_Conflicts"
        ]:
            for row in worksheet.iter_rows(min_row=2):
                for cell in row:
                    cell.fill = exception_fill
                    cell.font = Font(
                        color="000000",
                        bold=True
                    )


# ==========================================================
# 38. FINAL VALIDATION
# ==========================================================

assert (
    len(claim_item_summary)
    ==
    claim_item_summary[
        "EntityClaimItemEID"
    ].nunique()
), (
    "Validation failed: claim_item_summary "
    "does not contain one row per EntityClaimItemEID."
)

assert (
    candidate_category_total
    ==
    len(claim_item_summary)
), (
    "Validation failed: candidate categories "
    "do not reconcile to total claim items."
)

assert (
    claim_item_summary[
        "EntityClaimItemEID"
    ].notna().all()
), (
    "Validation failed: final analytical layer "
    "contains missing EntityClaimItemEID values."
)


# ==========================================================
# 39. PRINT RUN SUMMARY
# ==========================================================

print("=" * 65)
print(f"{OUTPUT_FILE} created successfully")
print("=" * 65)

print(
    "Input full_chain rows:",
    f"{input_row_count:,}"
)

print(
    "Unique claim items:",
    f"{total_analytical_claim_items:,}"
)

print(
    "No candidate matches:",
    f"{no_candidate_count:,}"
)

print(
    "Single candidate matches:",
    f"{single_candidate_count:,}"
)

print(
    "Multiple candidate matches:",
    f"{multiple_candidate_count:,}"
)

print(
    "Resolved single-candidate awards:",
    f"{resolved_award_count:,}"
)

print(
    "Valid cycle-time items:",
    f"{valid_cycle_time_count:,}"
)

print(
    "Negative cycle-time exceptions:",
    f"{negative_cycle_time_count:,}"
)

print(
    "Claim attribute conflict items:",
    f"{len(claim_attribute_conflicts):,}"
)

print(
    "Single-candidate award conflict items:",
    f"{len(single_candidate_conflicts):,}"
)

print(
    "Rows missing EntityClaimItemEID:",
    f"{missing_claim_item_id_rows:,}"
)

print("=" * 65)
```

```python
import pandas as pd
import numpy as np

# ==========================================================
# 1. DATE CLEANUP
# ==========================================================

date_cols = [
    "ClaimSubmittedDate",
    "AwardDate",
    "FulfillmentDate"
]

for col in date_cols:
    if col in full_chain.columns:
        full_chain[col] = pd.to_datetime(
            full_chain[col],
            errors="coerce"
        )

# ==========================================================
# 2. CLAIM ITEM ANALYTICAL LAYER
#    ONE ROW = ONE EntityClaimItemEID
# ==========================================================

claim_item_summary = (
    full_chain
    .groupby("EntityClaimItemEID")
    .agg(
        ClaimEId=("ClaimEId","first"),

        ParticipantId=("ParticipantId","first"),

        ParticipantName=(
            "ParticipantName",
            "first"
        ),

        EntityId=("EntityId","first"),

        PSPS=("PSPS","first"),

        PromotionName=(
            "PromotionName_claim",
            "first"
        ),

        ClaimItemStatus=(
            "ClaimItemStatus",
            "first"
        ),

        ClaimSubmittedDate=(
            "ClaimSubmittedDate",
            "first"
        ),

        AwardDate=(
            "AwardDate",
            "first"
        ),

        AwardAmount=(
            "AwardAmount",
            "first"
        ),

        DistinctLedgerIds=(
            "LedgerId",
            lambda x: x.dropna().nunique()
        )
    )
    .reset_index()
)

# ==========================================================
# 3. PAYOUT STATUS
# ==========================================================

claim_item_summary["PayoutStatus"] = np.select(
    [
        claim_item_summary["DistinctLedgerIds"] == 0,
        claim_item_summary["DistinctLedgerIds"] == 1,
        claim_item_summary["DistinctLedgerIds"] > 1
    ],
    [
        "No Candidate",
        "Single Candidate",
        "Multiple Candidates"
    ],
    default="Unknown"
)

# ==========================================================
# 4. CYCLE TIME METRICS
# ==========================================================

claim_item_summary["Days_Claim_to_Award"] = (
    claim_item_summary["AwardDate"]
    -
    claim_item_summary["ClaimSubmittedDate"]
).dt.days

cycle_summary = (
    claim_item_summary[
        ["Days_Claim_to_Award"]
    ]
    .describe()
)

# ==========================================================
# 5. PROMOTION DELAYS
# ==========================================================

promo_delay = (
    claim_item_summary
    .groupby("PromotionName")
    .agg(
        ClaimItems=("EntityClaimItemEID","nunique"),
        AvgDays=("Days_Claim_to_Award","mean"),
        MedianDays=("Days_Claim_to_Award","median"),
        MaxDays=("Days_Claim_to_Award","max")
    )
    .sort_values(
        "AvgDays",
        ascending=False
    )
)

# ==========================================================
# 6. ENTITY DELAYS
# ==========================================================

entity_delay = (
    claim_item_summary
    .groupby("EntityId")
    .agg(
        ClaimItems=("EntityClaimItemEID","nunique"),
        AvgDays=("Days_Claim_to_Award","mean"),
        MedianDays=("Days_Claim_to_Award","median"),
        MaxDays=("Days_Claim_to_Award","max")
    )
    .sort_values(
        "AvgDays",
        ascending=False
    )
)

# ==========================================================
# 7. PAYOUT DISTRIBUTION
# ==========================================================

payout_distribution = (
    claim_item_summary["PayoutStatus"]
    .value_counts()
    .reset_index()
)

payout_distribution.columns = [
    "PayoutStatus",
    "ClaimItems"
]

payout_distribution["Percent"] = (
    payout_distribution["ClaimItems"]
    /
    len(claim_item_summary)
    * 100
).round(2)

# ==========================================================
# 8. MULTIPLE PAYOUTS
# ==========================================================

multiple_payouts = (
    claim_item_summary[
        claim_item_summary["DistinctLedgerIds"] > 1
    ]
    .sort_values(
        "DistinctLedgerIds",
        ascending=False
    )
)

# ==========================================================
# 9. PROMO SPLIT PAYOUTS
# ==========================================================

promo_split = (
    claim_item_summary
    .groupby("PromotionName")
    .agg(
        ClaimItems=("EntityClaimItemEID","nunique"),

        MultipleCandidateItems=(
            "PayoutStatus",
            lambda x: (x == "Multiple Candidates").sum()
        )
    )
    .reset_index()
)

promo_split["SplitRatePct"] = (
    promo_split["MultipleCandidateItems"]
    /
    promo_split["ClaimItems"]
    * 100
).round(2)

promo_split = promo_split.sort_values(
    "SplitRatePct",
    ascending=False
)

# ==========================================================
# 10. AWARD VS POINTS
# ==========================================================
#
# Since candidate payouts are unresolved,
# focus on payout count behavior instead
#
# ==========================================================

award_vs_points = (
    claim_item_summary[
        [
            "EntityClaimItemEID",
            "ClaimEId",
            "AwardAmount",
            "DistinctLedgerIds",
            "PayoutStatus"
        ]
    ]
    .sort_values(
        "DistinctLedgerIds",
        ascending=False
    )
)

# ==========================================================
# 11. FULFILLMENT BUCKETS
# ==========================================================

claim_item_summary["FulfillmentBucket"] = pd.cut(
    claim_item_summary["Days_Claim_to_Award"],
    bins=[-999,7,30,60,90,9999],
    labels=[
        "0-7 Days",
        "8-30 Days",
        "31-60 Days",
        "61-90 Days",
        "90+ Days"
    ]
)

fulfillment_distribution = (
    claim_item_summary["FulfillmentBucket"]
    .value_counts()
    .reset_index()
)

fulfillment_distribution.columns = [
    "Bucket",
    "Count"
]

# ==========================================================
# 12. TOP PARTICIPANTS
# ==========================================================

participant_analysis = (
    claim_item_summary
    .groupby("ParticipantId")
    .agg(
        ClaimItems=("EntityClaimItemEID","nunique")
    )
    .sort_values(
        "ClaimItems",
        ascending=False
    )
)

# ==========================================================
# 13. CLAIMS NO PAYOUT
# ==========================================================

no_payout = (
    claim_item_summary[
        claim_item_summary["PayoutStatus"]
        ==
        "No Candidate"
    ]
)

no_payout_summary = (
    no_payout
    .groupby("PromotionName")
    .agg(
        ClaimItems=(
            "EntityClaimItemEID",
            "nunique"
        ),

        Claims=(
            "ClaimEId",
            "nunique"
        ),

        Participants=(
            "ParticipantId",
            "nunique"
        )
    )
    .reset_index()
    .sort_values(
        "ClaimItems",
        ascending=False
    )
)

# ==========================================================
# 14. EXECUTIVE KPIs
# ==========================================================

kpis = pd.DataFrame(
    {
        "Metric": [
            "Unique Claims",
            "Unique Claim Items",
            "Unique Participants",
            "Distinct Ledger IDs",
            "No Candidate Claim Items",
            "Single Candidate Claim Items",
            "Multiple Candidate Claim Items"
        ],
        "Value": [
            claim_item_summary[
                "ClaimEId"
            ].nunique(),

            claim_item_summary[
                "EntityClaimItemEID"
            ].nunique(),

            claim_item_summary[
                "ParticipantId"
            ].nunique(),

            full_chain[
                "LedgerId"
            ].nunique(),

            (
                claim_item_summary[
                    "PayoutStatus"
                ] == "No Candidate"
            ).sum(),

            (
                claim_item_summary[
                    "PayoutStatus"
                ] == "Single Candidate"
            ).sum(),

            (
                claim_item_summary[
                    "PayoutStatus"
                ] == "Multiple Candidates"
            ).sum()
        ]
    }
)

# ==========================================================
# 15. SAVE REPORT
# ==========================================================

with pd.ExcelWriter(
    "Business_Insights_Report_v2.xlsx",
    engine="openpyxl"
) as writer:

    kpis.to_excel(
        writer,
        sheet_name="Executive_KPIs",
        index=False
    )

    cycle_summary.to_excel(
        writer,
        sheet_name="Cycle_Time_Summary"
    )

    promo_delay.to_excel(
        writer,
        sheet_name="Promotion_Delays"
    )

    entity_delay.to_excel(
        writer,
        sheet_name="Entity_Delays"
    )

    payout_distribution.to_excel(
        writer,
        sheet_name="Payout_Distribution",
        index=False
    )

    multiple_payouts.to_excel(
        writer,
        sheet_name="Multiple_Payouts",
        index=False
    )

    promo_split.to_excel(
        writer,
        sheet_name="Promo_Split_Payouts",
        index=False
    )

    award_vs_points.to_excel(
        writer,
        sheet_name="Award_vs_Points",
        index=False
    )

    fulfillment_distribution.to_excel(
        writer,
        sheet_name="Fulfillment_Buckets",
        index=False
    )

    participant_analysis.to_excel(
        writer,
        sheet_name="Top_Participants"
    )

    no_payout_summary.to_excel(
        writer,
        sheet_name="Claims_No_Payout",
        index=False
    )

print("Business_Insights_Report_v2.xlsx created")
```
